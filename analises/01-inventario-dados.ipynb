# 01 — Inventário dos dados da CVM

**Objetivo:** descobrir o que existe dentro dos arquivos baixados da CVM.
**Entrada:** dados/bronze/cvm-inf-trimestral/*.zip
**Saída:** lista de CSVs e valores de Segmento_Atuacao
**Onde roda:** local (.venv) — **Camada:** Bronze

```python
# Importa as três ferramentas que vamos usar:
# zipfile para abrir arquivos .zip sem precisar extrair
# pandas para ler e manipular tabelas
# Path para trabalhar com caminhos de pasta de forma segura
import zipfile
import pandas as pd
from pathlib import Path
```

```python
# Descobre a pasta raiz do projeto.
# ".." significa "suba um nível" — saímos de analises/ e chegamos em tcc-galpoes/
# .resolve() converte esse caminho relativo no caminho completo e absoluto
RAIZ = Path("..").resolve()

# A partir da raiz, desce até a pasta dos ZIPs do informe trimestral.
# Com Path, a barra "/" não é divisão: ela junta pedaços de caminho
BRONZE_TRI = RAIZ / "dados" / "bronze" / "cvm-inf-trimestral"

# Imprime para conferirmos visualmente se os caminhos estão corretos
print("Raiz do projeto:", RAIZ)
print("Pasta dos ZIPs:", BRONZE_TRI)
print("Essa pasta existe?", BRONZE_TRI.exists())
```

```python
# Monta o caminho até um ZIP específico.
# Começamos com UM arquivo só, para entender a estrutura antes de repetir para todos.
arquivo_zip = BRONZE_TRI / "inf_trimestral_fii_2025.zip"

# Confirma que o arquivo existe antes de tentar abrir
print("Arquivo existe?", arquivo_zip.exists())

# Abre o ZIP em modo leitura. O "with" garante que ele será fechado
# automaticamente ao final, mesmo se algo der errado no meio do caminho.
with zipfile.ZipFile(arquivo_zip) as z:
    # namelist() devolve a lista de todos os arquivos que estão dentro do ZIP
    for nome in z.namelist():
        print(nome)
    
```

```python
# Nome exato do CSV que queremos ler, copiado da lista que acabamos de ver
nome_csv_geral = "inf_trimestral_fii_geral_2025.csv"

# Abre o ZIP e, de dentro dele, abre apenas o arquivo "geral" — sem extrair nada pro disco
with zipfile.ZipFile(arquivo_zip) as z:
    with z.open(nome_csv_geral) as f:
        # Lê o CSV e transforma em uma tabela (DataFrame).
        # sep=';' e encoding='latin-1' são o padrão dos arquivos da CVM
        df_geral = pd.read_csv(f, sep=';', encoding='latin-1')

# Mostra quantas linhas e colunas a tabela tem
print("Formato da tabela (linhas, colunas):", df_geral.shape)

# Mostra as primeiras linhas, só para conferir visualmente que carregou certo
df_geral.head()
```

```python
# .unique() devolve todos os valores DISTINTOS que aparecem na coluna,
# sem repetição — é assim que descobrimos quais segmentos existem
segmentos = df_geral['Segmento_Atuacao'].unique()

# Imprime a lista de segmentos encontrados
print("Segmentos encontrados:")
for s in segmentos:
    print("-", s)
```

```python
# Filtra o DataFrame, mantendo só as linhas onde Segmento_Atuacao é exatamente "Logística"
df_logistica = df_geral[df_geral['Segmento_Atuacao'] == 'Logística']

# Quantas linhas sobraram após o filtro?
print("Linhas com segmento Logística:", df_logistica.shape[0])

# CNPJs distintos — um fundo pode aparecer mais de uma vez
cnpjs_logisticos = df_logistica['CNPJ_Fundo_Classe'].unique()
print("Quantidade de fundos logísticos distintos:", len(cnpjs_logisticos))

# Nomes dos fundos, para conferência visual
nomes_logisticos = df_logistica['Nome_Fundo_Classe'].unique()
print("\nFundos encontrados:")
for nome in nomes_logisticos:
    print("-", nome)
```
